# Michaelis–Menten Kinetics: Data, Model, and Inhibition

This notebook has three parts:

1. **Data**
2. **Michaelis–Menten**
3. **Michaelis–Menten + inhibition**

Run each cell from top to bottom. The goal is to connect the data to the Michaelis–Menten model, then see how an inhibitor changes the kinetic curve.


## 1. Data

Start with measured initial velocities at several substrate concentrations.


```python
import numpy as np
import matplotlib.pyplot as plt

# Substrate concentrations [S] (mM) and initial velocities v0 (μM/min)
S = np.array([0.5, 1, 2, 4, 8, 16], dtype=float)
v0 = np.array([0.05, 0.09, 0.16, 0.26, 0.35, 0.42], dtype=float)

plt.scatter(S, v0)
plt.xlabel('[S] (mM)')
plt.ylabel('Initial velocity, v0 (μM/min)')
plt.title('Enzyme kinetics data')
plt.show()
```

## 2. Michaelis–Menten

For a simple Michaelis–Menten enzyme,

\[
v_0 = \frac{V_{\max}[S]}{K_M + [S]}
\]

Fit the Michaelis–Menten equation directly to the data and compare the model with the measurements.


```python
from scipy.optimize import curve_fit

def michaelis_menten(S, Vmax, Km):
    return Vmax * S / (Km + S)

# Fit Vmax and Km to the data
popt, pcov = curve_fit(michaelis_menten, S, v0, p0=[0.5, 2.0])
Vmax_fit, Km_fit = popt

print(f'Vmax = {Vmax_fit:.3f} μM/min')
print(f'Km   = {Km_fit:.3f} mM')

S_smooth = np.linspace(0, 18, 300)

plt.scatter(S, v0, label='data')
plt.plot(S_smooth, michaelis_menten(S_smooth, Vmax_fit, Km_fit), label='Michaelis–Menten fit')
plt.xlabel('[S] (mM)')
plt.ylabel('Initial velocity, v0 (μM/min)')
plt.title('Michaelis–Menten model')
plt.legend()
plt.show()
```

## 3. Michaelis–Menten + inhibition

Now add a **reversible inhibitor** to the Michaelis–Menten model.

For the example below, the inhibitor competes with substrate for free enzyme:

\[
v_0 = \frac{V_{\max}[S]}{K_M\left(1+\frac{[I]}{K_I}\right)+[S]}
\]

**You do not need to derive or fit this equation here.** Just look at how increasing inhibitor concentration changes the relationship between \(v_0\) and \([S]\).


```python
def michaelis_menten_with_inhibition(S, Vmax, Km, I, Ki):
    return Vmax * S / (Km * (1 + I / Ki) + S)

# Example inhibition parameters
Ki = 2.0  # mM
I_values = [0, 1, 2, 4]  # mM

for I in I_values:
    v_inhibited = michaelis_menten_with_inhibition(S_smooth, Vmax_fit, Km_fit, I, Ki)
    plt.plot(S_smooth, v_inhibited, label=f'[I] = {I} mM')

plt.xlabel('[S] (mM)')
plt.ylabel('Initial velocity, v0 (μM/min)')
plt.title('Michaelis–Menten kinetics with inhibition')
plt.legend()
plt.show()

```
