## Step 1: Dataset & Raw Plot

We'll start by loading a small enzyme kinetics dataset.

```
import numpy as np
import matplotlib.pyplot as plt

# Example substrate concentrations [S] (mM) and velocities V (μM/min)
S = np.array([0.5, 1, 2, 4, 8, 16], dtype=float)
V = np.array([0.05, 0.09, 0.16, 0.26, 0.35, 0.42], dtype=float)

plt.scatter(S, V, color='blue')
plt.xlabel('[S] (mM)')
plt.ylabel('V (μM/min)')
plt.title('Raw Michaelis–Menten Data')
plt.show()
```

## Step 2: Lineweaver–Burk (Worked Example)

The Lineweaver–Burk transformation linearizes the Michaelis–Menten equation:
\[
\frac{1}{V} = \frac{K_M}{V_{\max}} \cdot \frac{1}{[S]} + \frac{1}{V_{\max}}
\]
Here, the slope = $K_M / V_{max}$ and intercept = $1/V_{max}$.

```
from scipy.stats import linregress

# Transform data
inv_S = 1/S
inv_V = 1/V

# Fit line
slope, intercept, r_value, p_value, std_err = linregress(inv_S, inv_V)

# Calculate parameters
Vmax_LB = 1/intercept
Km_LB = slope * Vmax_LB

print(f'Lineweaver–Burk: Vmax = {Vmax_LB:.3f}, Km = {Km_LB:.3f}')

# Plot
plt.scatter(inv_S, inv_V, color='blue')
plt.plot(inv_S, slope*inv_S + intercept, color='red')
plt.xlabel('1/[S] (1/mM)')
plt.ylabel('1/V (min/μM)')
plt.title('Lineweaver–Burk Plot')
plt.show()
```
